```python
def extract_text(filepath):
    ext = Path(filepath).suffix.lower()

    # ── PDF ──────────────────────────────────────────────────
    if ext == ".pdf":
        text = ""
        try:
            with pdfplumber.open(filepath) as pdf:
                for page in pdf.pages:
                    t = page.extract_text()
                    if t:
                        text += t + "\n"
        except Exception as e:
            print(f"  pdfplumber error: {e}")

        # If text too short → it's a scanned image, use OCR
        if len(text.strip()) < 100:
            print("  → Switching to OCR (scanned document detected)")
            try:
                images = convert_from_path(filepath, dpi=300)
                for img in images:
                    text += pytesseract.image_to_string(img) + "\n"
            except Exception as e:
                print(f"  OCR error: {e}")
        return text.strip()

    # ── DOCX ─────────────────────────────────────────────────
    elif ext in [".docx", ".doc"]:
        from docx import Document
        doc = Document(filepath)
        return "\n".join([p.text for p in doc.paragraphs]).strip()

    # ── IMAGE ────────────────────────────────────────────────
    elif ext in [".png", ".jpg", ".jpeg", ".tiff", ".bmp"]:
        img = Image.open(filepath)
        return pytesseract.image_to_string(img).strip()

    else:
        return ""

print("✅ Text extractor ready!")
```

```python
def classify_document(text):
    text_lower = text.lower()
    scores = {}

    for doc_type, template in TEMPLATES.items():
        score = 0
        for keyword in template["keywords"]:
            if keyword.lower() in text_lower:
                score += 1
            else:
                # fuzzy match for typos/OCR errors
                for line in text_lower.split('\n'):
                    if fuzz.partial_ratio(keyword.lower(), line) > 85:
                        score += 0.5
                        break
        scores[doc_type] = score / len(template["keywords"])

    best = max(scores, key=scores.get)
    confidence = round(scores[best] * 100, 1)

    if confidence < 20:
        return "UNKNOWN", 0.0

    return best, confidence

print("✅ Classifier ready!")
```

```python
def extract_fields(text):
    text_upper = text.upper()
    fields = {}

    # TIN / NIF numbers
    tins = re.findall(r'(?:TIN|NIF)\s*(?:NUMBER)?\s*[:\-]?\s*(\d{6,12})', text_upper)
    fields['tin_numbers'] = list(set(tins))

    # BL / Bill of Lading numbers
    bls = re.findall(r'(?:BILL OF LAD(?:ING|ING NO)|BILL OF LOAD(?:ING)?|BL NUMBER|BL NO|B/L)[:\s#]*([A-Z]{2,4}\d{6,10})', text_upper)
    fields['bl_numbers'] = list(set(bls))

    # Container number
    container = re.search(r'CONTAINER\s*(?:NO|NUMBER|N[O°]?)[:\s]*([A-Z]{4}\d{7})', text_upper)
    fields['container'] = container.group(1) if container else None

    # Vessel name
    vessel = re.search(r'(?:NAME OF VESSEL|VESSEL\s*VOYAGE)[:\s]*([\w\s]+?)(?:\n|/)', text_upper)
    fields['vessel'] = vessel.group(1).strip() if vessel else None

    # Voyage number
    voyage = re.search(r'VOYAGE\s*(?:NUMBER)?[:\s]*(\w+)', text_upper)
    fields['voyage'] = voyage.group(1).strip() if voyage else None

    # Consignee
    consignee = re.search(r'CONSIGN(?:EE|E)[:\s]+([^\n]+)', text_upper)
    fields['consignee'] = consignee.group(1).strip() if consignee else None

    # Invoice number
    inv = re.search(r'INVOICE\s*(?:NO\.?|NUMBER)[:\s]+(\S+)', text_upper)
    fields['invoice_no'] = inv.group(1).strip() if inv else None

    # Dates
    dates = re.findall(r'\b(\d{1,2}[/\-]\d{1,2}[/\-]\d{2,4})\b', text)
    fields['dates'] = list(set(dates))

    # Total amount
    total = re.search(r'TOTAL\s*[/USD]*[:\s]*(\d[\d,\.]+)', text_upper)
    fields['total'] = float(total.group(1).replace(',', '')) if total else None

    # Origin / Destination
    origin = re.search(r'ORIGIN\s*[:\s]+([^\n]+)', text_upper)
    fields['origin'] = origin.group(1).strip() if origin else None

    dest = re.search(r'(?:FINAL\s+)?DESTINATION\s*[:\s]+([^\n]+)', text_upper)
    fields['destination'] = dest.group(1).strip() if dest else None

    # Weight
    weight = re.search(r'WEIGHT\s*[:\s]*(\d+\s*KGS?)', text_upper)
    fields['weight'] = weight.group(1).strip() if weight else None

    return fields

print("✅ Field extractor ready!")
```

```python
def check_document(text, doc_type, fields):
    issues = []
    text_upper = text.upper()
    template = TEMPLATES.get(doc_type)

    if not template:
        issues.append({"severity": "CRITICAL", "check": "Type",
                       "message": "Document type unknown — does not match any template"})
        return issues

    # ── 1. TIN NUMBER CHECK ──────────────────────────────────
    valid_tins = template.get("valid_tins", [])
    found_tins = fields.get("tin_numbers", [])

    if valid_tins:
        if not found_tins:
            issues.append({"severity": "HIGH", "check": "TIN",
                           "message": f"TIN number missing. Expected: {valid_tins}"})
        else:
            for tin in found_tins:
                if tin not in ALL_VALID_TINS:
                    issues.append({"severity": "CRITICAL", "check": "TIN",
                                   "message": f"🚨 FRAUD: TIN '{tin}' is NOT in the valid TIN list!"})
                elif tin not in valid_tins:
                    issues.append({"severity": "HIGH", "check": "TIN",
                                   "message": f"TIN '{tin}' is valid globally but wrong for this document type"})

    # ── 2. REQUIRED FIELDS CHECK ─────────────────────────────
    missing = []
    for field in template.get("required_fields", []):
        in_text = field.upper() in text_upper
        fuzzy_match = any(fuzz.partial_ratio(field.upper(), line) > 80
                          for line in text_upper.split('\n'))
        if not in_text and not fuzzy_match:
            missing.append(field)

    if missing:
        issues.append({"severity": "MEDIUM", "check": "Required Fields",
                       "message": f"Missing: {', '.join(missing)}"})

    # ── 3. BL NUMBER CHECK ───────────────────────────────────
    for bl in fields.get("bl_numbers", []):
        if bl not in KNOWN_BL_NUMBERS:
            issues.append({"severity": "HIGH", "check": "BL Number",
                           "message": f"BL '{bl}' not in known valid BLs: {KNOWN_BL_NUMBERS}"})

    # ── 4. CONTAINER CHECK ───────────────────────────────────
    container = fields.get("container")
    if container and container not in KNOWN_CONTAINERS:
        issues.append({"severity": "HIGH", "check": "Container",
                       "message": f"Container '{container}' not in known containers: {KNOWN_CONTAINERS}"})

    # ── 5. VESSEL CHECK ──────────────────────────────────────
    vessel = fields.get("vessel")
    if vessel:
        ok = any(fuzz.partial_ratio(vessel, v) > 80 for v in KNOWN_VESSELS)
        if not ok:
            issues.append({"severity": "MEDIUM", "check": "Vessel",
                           "message": f"Vessel '{vessel}' not matching known vessels"})

    # ── 6. VOYAGE CHECK ──────────────────────────────────────
    voyage = fields.get("voyage")
    if voyage and voyage not in KNOWN_VOYAGES:
        issues.append({"severity": "MEDIUM", "check": "Voyage",
                       "message": f"Voyage '{voyage}' not in known voyages: {KNOWN_VOYAGES}"})

    # ── 7. ISSUER / COMPANY NAME CHECK ───────────────────────
    issuer_found = any(
        fuzz.partial_ratio(issuer.upper(), text_upper) > 75
        for issuer in template.get("valid_issuers", [])
    )
    if not issuer_found:
        issues.append({"severity": "HIGH", "check": "Issuer",
                       "message": f"Expected company name not found: {template['valid_issuers']}"})

    # ── 8. SIGNATURE CHECK ───────────────────────────────────
    for sig in template.get("required_signatures", []):
        if sig.upper() not in text_upper:
            issues.append({"severity": "MEDIUM", "check": "Signature",
                           "message": f"Missing signature field: '{sig}'"})

    # ── 9. AI-GENERATED CONTENT CHECK ────────────────────────
    ai_signals = ["as an ai", "i am an ai", "language model", "chatgpt",
                  "openai", "i cannot assist", "artificial intelligence assistant"]
    for signal in ai_signals:
        if signal in text.lower():
            issues.append({"severity": "CRITICAL", "check": "AI Content",
                           "message": f"🚨 FRAUD: AI-generated text detected — '{signal}'"})

    # ── 10. DATE PRESENT CHECK ───────────────────────────────
    if not fields.get("dates"):
        issues.append({"severity": "MEDIUM", "check": "Date",
                       "message": "No date found in document"})

    return issues

print("✅ Fraud checker ready!")
```

```python
def cross_check(all_results):
    """Check that multiple documents in the same shipment are consistent."""
    issues = []

    bl_map        = {}
    container_map = {}
    voyage_map    = {}

    for r in all_results:
        fname = r['filename']
        fields = r['fields']
        for bl in fields.get('bl_numbers', []):
            bl_map.setdefault(bl, []).append(fname)
        if fields.get('container'):
            container_map.setdefault(fields['container'], []).append(fname)
        if fields.get('voyage'):
            voyage_map.setdefault(fields['voyage'], []).append(fname)

    if len(bl_map) > 1:
        issues.append({"severity": "HIGH", "check": "BL Consistency",
                       "message": f"Multiple BL numbers across docs: {list(bl_map.keys())}"})

    if len(container_map) > 1:
        issues.append({"severity": "MEDIUM", "check": "Container Consistency",
                       "message": f"Multiple container numbers: {list(container_map.keys())}"})

    if len(voyage_map) > 1:
        issues.append({"severity": "MEDIUM", "check": "Voyage Consistency",
                       "message": f"Multiple voyage numbers: {list(voyage_map.keys())}"})

    return issues

print("✅ Cross-check ready!")
```

```python
def audit(filepath):
    filename = Path(filepath).name
    print(f"\n{'─'*55}")
    print(f"📄  {filename}")
    print(f"{'─'*55}")

    # Extract text
    text = extract_text(filepath)
    if not text:
        print("  ❌ Could not extract text")
        return {'filename': filename, 'status': 'ERROR', 'issues': [], 'fields': {}, 'doc_type': 'UNKNOWN'}

    print(f"  Characters extracted: {len(text)}")

    # Classify
    doc_type, confidence = classify_document(text)
    doc_name = TEMPLATES.get(doc_type, {}).get('name', doc_type)
    print(f"  Type: {doc_name}  ({confidence}% confidence)")

    # Extract fields
    fields = extract_fields(text)
    print(f"  TINs: {fields.get('tin_numbers')}  |  BLs: {fields.get('bl_numbers')}  |  Container: {fields.get('container')}")

    # Check
    issues = check_document(text, doc_type, fields)

    # Verdict
    critical = [i for i in issues if i['severity'] == 'CRITICAL']
    high     = [i for i in issues if i['severity'] == 'HIGH']
    medium   = [i for i in issues if i['severity'] == 'MEDIUM']

    if critical:
        status  = 'REJECTED'
        verdict = '🚨 REJECTED — Critical issues found'
    elif high:
        status  = 'FLAGGED'
        verdict = '⚠️  FLAGGED — Needs manual review'
    elif medium:
        status  = 'WARNING'
        verdict = '🟡 APPROVED WITH WARNINGS'
    else:
        status  = 'APPROVED'
        verdict = '✅ APPROVED — All checks passed'

    print(f"\n  {verdict}")

    if issues:
        icons = {'CRITICAL': '🚨', 'HIGH': '⚠️ ', 'MEDIUM': '🟡'}
        for i in issues:
            print(f"  {icons.get(i['severity'], 'ℹ️ ')} [{i['severity']}] {i['check']}: {i['message']}")

    return {
        'filename':     filename,
        'doc_type':     doc_type,
        'doc_type_name':doc_name,
        'confidence':   confidence,
        'status':       status,
        'verdict':      verdict,
        'fields':       fields,
        'issues':       issues,
        'critical':     len(critical),
        'high':         len(high),
        'medium':       len(medium),
    }

print("✅ Audit function ready!")
```

```python
def generate_report(results, cross_issues):
    summary = []
    detail  = []

    for r in results:
        summary.append({
            'File':           r['filename'],
            'Type':           r.get('doc_type_name', ''),
            'Confidence':     f"{r.get('confidence', 0)}%",
            'Status':         r['status'],
            'Critical':       r.get('critical', 0),
            'High':           r.get('high', 0),
            'Medium':         r.get('medium', 0),
            'TINs':           ', '.join(r['fields'].get('tin_numbers', [])),
            'BL Numbers':     ', '.join(r['fields'].get('bl_numbers', [])),
            'Container':      r['fields'].get('container', ''),
            'Vessel':         r['fields'].get('vessel', ''),
            'Voyage':         r['fields'].get('voyage', ''),
            'Consignee':      r['fields'].get('consignee', ''),
            'Verdict':        r['verdict'],
        })
        for issue in r['issues']:
            detail.append({
                'File':     r['filename'],
                'Severity': issue['severity'],
                'Check':    issue['check'],
                'Issue':    issue['message'],
            })

    cross_rows = [{'Severity': i['severity'], 'Check': i['check'],
                   'Issue': i['message']} for i in cross_issues]

    path = "SIFCO_Audit_Report.xlsx"
    with pd.ExcelWriter(path, engine='openpyxl') as writer:
        pd.DataFrame(summary).to_excel(writer, sheet_name='Summary', index=False)
        if detail:
            pd.DataFrame(detail).to_excel(writer, sheet_name='Issues', index=False)
        if cross_rows:
            pd.DataFrame(cross_rows).to_excel(writer, sheet_name='Cross-Doc Issues', index=False)

    print(f"\n📊 Report saved: {path}")
    files.download(path)
    return path

print("✅ Report generator ready!")
```

```python
# Upload your documents
print("📁 Upload your documents...")
uploaded = files.upload()

results = []
for filepath in uploaded.keys():
    result = audit(filepath)
    results.append(result)

# Cross-document check
print(f"\n{'─'*55}")
print("🔗 Cross-document consistency check...")
cross_issues = cross_check(results)

if cross_issues:
    for i in cross_issues:
        print(f"  ⚠️  [{i['severity']}] {i['check']}: {i['message']}")
else:
    print("  ✅ All documents consistent with each other")

# Final summary
print(f"\n{'═'*55}")
print("📋  FINAL SUMMARY")
print(f"{'═'*55}")
print(f"  Total:    {len(results)}")
print(f"  ✅ Approved: {sum(1 for r in results if r['status'] == 'APPROVED')}")
print(f"  🟡 Warnings: {sum(1 for r in results if r['status'] == 'WARNING')}")
print(f"  ⚠️  Flagged:  {sum(1 for r in results if r['status'] == 'FLAGGED')}")
print(f"  🚨 Rejected: {sum(1 for r in results if r['status'] == 'REJECTED')}")
```
